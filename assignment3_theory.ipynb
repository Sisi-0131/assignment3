```python
# Q5:
# Given a mini-batch of 4 values: [6, 8, 10, 6], 
# compute the normalized output using BatchNorm without γ and β (i.e., just zero-mean, unit-variance).

import torch

values = torch.tensor(
    [6.0, 8.0, 10.0, 6.0]
)

mean = values.mean()

var = values.var(
    unbiased=False
)

sd = torch.sqrt(
    var
)

normalized_values = (
    values - mean
) / sd

formatted_output = [
    f"{value.item():.2f}"
    for value in normalized_values
]

print("Original values:", values.tolist())
print(f"Mean: {mean.item():.2f}")
print(f"Variance: {var.item():.2f}")
print(
    f"Standard deviation: "
    f"{sd.item():.2f}"
)
print(
    "Normalized output: ["
    + ", ".join(formatted_output)
    + "]"
)
```

```python
# Q6:
# What is the key mathematical difference between ReLU and LeakyReLU? Give their formulas
import torch
import torch.nn.functional as F

x = torch.tensor(
    [-5.0, 0.0, 5.0]
)

relu_output = F.relu(x)

leaky_relu_output = F.leaky_relu(
    x,
    negative_slope=0.2
)

print("Input:", x.tolist())
print(
    "ReLU output:",
    relu_output.tolist()
)
print(
    "LeakyReLU output:",
    leaky_relu_output.tolist()
)
```
